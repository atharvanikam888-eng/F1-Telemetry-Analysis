```python

# ==========================================
# F1 TYRE COMPOUND COMPARISON
# Abu Dhabi GP 2022
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt

# 1. Prepare data
compound_data = clean_laps.copy()

# Keep green-flag laps only
compound_data = compound_data[
    compound_data['TrackStatus'].astype(str) == '1'
].copy()

# Define each individual tyre stint
stint_columns = ['Driver', 'Stint', 'Compound']

# Sort laps by tyre age
compound_data = compound_data.sort_values(
    ['Driver', 'Stint', 'TyreLife']
)

# 2. Calculate baseline lap time for each stint
# Baseline = average of the first 3 recorded accurate laps

baseline = (
    compound_data
    .groupby(stint_columns)
    .head(3)
    .groupby(stint_columns)['LapTimeSeconds']
    .mean()
    .rename('BaselineLapTime')
    .reset_index()
)

# Keep only stints with at least 3 laps
stint_counts = (
    compound_data
    .groupby(stint_columns)
    .size()
    .reset_index(name='LapCount')
)

valid_stints = stint_counts[
    stint_counts['LapCount'] >= 3
][stint_columns]

compound_data = compound_data.merge(
    valid_stints,
    on=stint_columns,
    how='inner'
)

# 3. Calculate lap-time change from baseline
compound_data = compound_data.merge(
    baseline,
    on=stint_columns,
    how='inner'
)

compound_data['LapTimeChange'] = (
    compound_data['LapTimeSeconds']
    - compound_data['BaselineLapTime']
)

# 4. Calculate median lap-time change
compound_summary = (
    compound_data
    .groupby(['Compound', 'TyreLife'])
    .agg(
        MedianLapTimeChange=('LapTimeChange', 'median'),
        SampleCount=('LapTimeChange', 'count')
    )
    .reset_index()
)

# 5. Display results
print("TYRE COMPOUND PERFORMANCE COMPARISON")
print(compound_summary.to_string(index=False))

# 6. Plot comparison
plt.figure(figsize=(12, 7))

for compound in ['SOFT', 'MEDIUM', 'HARD']:

    data = compound_summary[
        compound_summary['Compound'] == compound
    ]

    if not data.empty:
        plt.plot(
            data['TyreLife'],
            data['MedianLapTimeChange'],
            marker='o',
            linewidth=2,
            label=compound
        )

plt.axhline(
    y=0,
    linestyle='--',
    linewidth=1
)

plt.xlabel('Tyre Age (Laps)')
plt.ylabel('Median Lap-Time Change from Stint Baseline (Seconds)')
plt.title('F1 Tyre Compound Performance - Abu Dhabi 2022')

plt.grid(True, alpha=0.3)
plt.legend(title='Tyre Compound')
plt.tight_layout()
plt.show()
```

```python

# ==========================================
# DATA COVERAGE AND RELIABILITY CHECK
# ==========================================

# Create a unique identifier for each driver's tyre stint
compound_data['StintID'] = (
    compound_data['Driver'].astype(str)
    + '_'
    + compound_data['Stint'].astype(str)
)

# Calculate sample coverage at each tyre age
coverage = (
    compound_data
    .groupby(['Compound', 'TyreLife'])
    .agg(
        LapSamples=('LapTimeSeconds', 'count'),
        Drivers=('Driver', 'nunique'),
        Stints=('StintID', 'nunique')
    )
    .reset_index()
)

print("SAMPLE COVERAGE BY TYRE COMPOUND")
print(coverage.to_string(index=False))

# ==========================================
# FILTER FOR BETTER-SUPPORTED RESULTS
# ==========================================

MIN_LAPS = 5
MIN_STINTS = 3

reliable_ages = coverage[
    (coverage['LapSamples'] >= MIN_LAPS) &
    (coverage['Stints'] >= MIN_STINTS)
][['Compound', 'TyreLife']]

reliable_data = compound_summary.merge(
    reliable_ages,
    on=['Compound', 'TyreLife'],
    how='inner'
)

# ==========================================
# REVISED COMPOUND COMPARISON
# ==========================================

plt.figure(figsize=(12, 7))

for compound in ['SOFT', 'MEDIUM', 'HARD']:

    data = reliable_data[
        reliable_data['Compound'] == compound
    ].sort_values('TyreLife')

    if not data.empty:
        plt.plot(
            data['TyreLife'],
            data['MedianLapTimeChange'],
            marker='o',
            linewidth=2,
            label=compound
        )

plt.axhline(y=0, linestyle='--', linewidth=1)

plt.xlabel('Tyre Age (Laps)')
plt.ylabel('Median Lap-Time Change from Stint Baseline (Seconds)')
plt.title('Tyre Compound Comparison - Reliability Filter Applied')

plt.grid(True, alpha=0.3)
plt.legend(title='Tyre Compound')
plt.tight_layout()
plt.show()
```

```python

print("RELIABILITY SUMMARY")

for compound in ['SOFT', 'MEDIUM', 'HARD']:

    data = reliable_data[
        reliable_data['Compound'] == compound
    ]

    if data.empty:
        print(f"\n{compound}: No tyre ages passed the filter.")
        continue

    print(f"\n{compound}")
    print(f"Reliable tyre age range: "
          f"{data['TyreLife'].min():.0f} to "
          f"{data['TyreLife'].max():.0f} laps")

    print(f"Number of reliable tyre ages: {len(data)}")

    print(f"Median lap-time change at final reliable age: "
          f"{data.sort_values('TyreLife')['MedianLapTimeChange'].iloc[-1]:.3f} seconds")

print("\nTOTAL RELIABLE DATA POINTS:", len(reliable_data))
```

```python

# ==========================================
# TYRE DEGRADATION RATE BY STINT
# ==========================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

stint_results = []

# Analyse each driver's individual tyre stint
for (driver, stint, compound), group in compound_data.groupby(
    ['Driver', 'Stint', 'Compound']
):

    group = group.sort_values('TyreLife')

    # Require at least 8 observations and 8 different tyre ages
    if len(group) < 8 or group['TyreLife'].nunique() < 8:
        continue

    # Calculate linear regression
    slope, intercept = np.polyfit(
        group['TyreLife'],
        group['LapTimeSeconds'],
        1
    )

    # Calculate R-squared
    predicted = (
        slope * group['TyreLife'] + intercept
    )

    actual = group['LapTimeSeconds']

    ss_res = ((actual - predicted) ** 2).sum()
    ss_tot = ((actual - actual.mean()) ** 2).sum()

    r_squared = (
        1 - ss_res / ss_tot
        if ss_tot != 0 else np.nan
    )

    stint_results.append({
        'Driver': driver,
        'Stint': stint,
        'Compound': compound,
        'LapsAnalysed': len(group),
        'TyreAgeRange': (
            group['TyreLife'].max()
            - group['TyreLife'].min()
        ),
        'DegradationRate': slope,
        'R2': r_squared
    })

# Convert results into a DataFrame
degradation_rates = pd.DataFrame(stint_results)

# Summarise results by compound
degradation_summary = (
    degradation_rates
    .groupby('Compound')
    .agg(
        NumberOfStints=('DegradationRate', 'count'),
        MedianRate=('DegradationRate', 'median'),
        MeanRate=('DegradationRate', 'mean'),
        LowerQuartile=('DegradationRate', lambda x: x.quantile(0.25)),
        UpperQuartile=('DegradationRate', lambda x: x.quantile(0.75)),
        MedianR2=('R2', 'median')
    )
    .round(4)
)

print("DEGRADATION RATE BY COMPOUND")
display(degradation_summary)

print("\nINDIVIDUAL STINT RESULTS")
display(degradation_rates.sort_values(
    ['Compound', 'DegradationRate']
))

# ==========================================
# VISUALISE DEGRADATION RATES
# ==========================================

plt.figure(figsize=(10, 6))

degradation_rates.boxplot(
    column='DegradationRate',
    by='Compound',
    grid=True
)

plt.axhline(y=0, linestyle='--', linewidth=1)

plt.title('Lap-Time Degradation Rate by Tyre Compound')
plt.suptitle('')
plt.xlabel('Tyre Compound')
plt.ylabel('Lap-Time Change (Seconds per Tyre-Age Lap)')

plt.tight_layout()
plt.show()
```

```python

# ==========================================
# FINAL COMPOUND DEGRADATION SUMMARY
# ==========================================

summary = (
    degradation_rates
    .groupby('Compound')
    .agg(
        StintsAnalysed=('DegradationRate', 'count'),
        MedianRate=('DegradationRate', 'median'),
        MeanRate=('DegradationRate', 'mean'),
        MinRate=('DegradationRate', 'min'),
        MaxRate=('DegradationRate', 'max'),
        MedianR2=('R2', 'median')
    )
    .round(4)
)

print("TYRE COMPOUND DEGRADATION SUMMARY")
display(summary)

# ==========================================
# BOXPLOT: DISTRIBUTION OF DEGRADATION RATES
# ==========================================

plt.figure(figsize=(10, 6))

data_to_plot = [
    degradation_rates[
        degradation_rates['Compound'] == compound
    ]['DegradationRate'].values
    for compound in ['SOFT', 'MEDIUM', 'HARD']
]

plt.boxplot(
    data_to_plot,
    tick_labels=['SOFT', 'MEDIUM', 'HARD'],
    showmeans=True
)

plt.axhline(0, linestyle='--', linewidth=1)

plt.xlabel('Tyre Compound')
plt.ylabel('Lap-Time Trend (Seconds per Tyre-Age Lap)')
plt.title('Distribution of Lap-Time Trends by Tyre Compound')

plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Create a unique identifier for every stint across both races
multi_race_laps['StintID'] = (
    multi_race_laps['Event'].astype(str) + '_' +
    multi_race_laps['Driver'].astype(str) + '_' +
    multi_race_laps['Stint'].astype(str)
)

# Sort laps in chronological order within each stint
multi_race_laps = multi_race_laps.sort_values(
    ['Event', 'Driver', 'Stint', 'TyreLife']
).copy()

# Calculate each stint's baseline using its first three recorded laps
baseline = (
    multi_race_laps
    .groupby('StintID')
    .head(3)
    .groupby('StintID')['LapTimeSeconds']
    .mean()
    .rename('BaselineLapTime')
)

multi_race_laps = multi_race_laps.join(
    baseline,
    on='StintID'
)

# Calculate lap-time change relative to the stint baseline
multi_race_laps['LapTimeChange'] = (
    multi_race_laps['LapTimeSeconds'] -
    multi_race_laps['BaselineLapTime']
)

# Summarise by race, compound and tyre age
race_compound_summary = (
    multi_race_laps
    .groupby(['Event', 'Compound', 'TyreLife'])
    .agg(
        MedianLapTimeChange=('LapTimeChange', 'median'),
        LapSamples=('LapTimeChange', 'size'),
        Stints=('StintID', 'nunique')
    )
    .reset_index()
)

# Keep only reasonably supported observations
reliable_summary = race_compound_summary[
    (race_compound_summary['LapSamples'] >= 5) &
    (race_compound_summary['Stints'] >= 3)
].copy()

print("DATASET SUMMARY")
print("Total laps:", len(multi_race_laps))
print("Total stints:", multi_race_laps['StintID'].nunique())

print("\nLAPS BY EVENT")
print(multi_race_laps.groupby('Event').size())

print("\nRELIABLE OBSERVATIONS")
print(
    reliable_summary.groupby(['Event', 'Compound']).agg(
        TyreAges=('TyreLife', 'nunique'),
        Stints=('Stints', 'max'),
        FinalMedianChange=('MedianLapTimeChange', 'last')
    ).round(3)
)

# Plot each circuit separately
fig, axes = plt.subplots(1, 2, figsize=(15, 5), sharey=True)

for ax, event in zip(axes, ['Abu Dhabi', 'Bahrain']):
    event_data = reliable_summary[
        reliable_summary['Event'] == event
    ]

    for compound in event_data['Compound'].unique():
        compound_data = event_data[
            event_data['Compound'] == compound
        ].sort_values('TyreLife')

        ax.plot(
            compound_data['TyreLife'],
            compound_data['MedianLapTimeChange'],
            marker='o',
            label=compound
        )

    ax.set_title(event)
    ax.set_xlabel('Tyre Age (Laps)')
    ax.grid(True, alpha=0.3)
    ax.legend()

axes[0].set_ylabel('Median Lap-Time Change (seconds)')
fig.suptitle('Tyre Performance Trends Across Two Circuits')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Prepare clean race laps for both drivers
comparison_laps = session.laps[
    session.laps['Driver'].isin(['VER', 'PER'])
].copy()

comparison_laps = comparison_laps[
    (comparison_laps['IsAccurate'] == True) &
    (comparison_laps['TrackStatus'].astype(str) == '1') &
    comparison_laps['LapTime'].notna() &
    comparison_laps['TyreLife'].notna()
].copy()

# Verstappen's reference stint
ver_reference = comparison_laps[
    (comparison_laps['Driver'] == 'VER') &
    (comparison_laps['Stint'] == 1)
].sort_values('TyreLife')

reference_compound = ver_reference['Compound'].mode().iloc[0]

ver_min_age = ver_reference['TyreLife'].min()
ver_max_age = ver_reference['TyreLife'].max()

# Find Pérez stints using the same compound
perez_laps = comparison_laps[
    comparison_laps['Driver'] == 'PER'
]

candidate_stints = []

for stint, group in perez_laps.groupby('Stint'):

    compound = group['Compound'].mode()

    if compound.empty or compound.iloc[0] != reference_compound:
        continue

    perez_min_age = group['TyreLife'].min()
    perez_max_age = group['TyreLife'].max()

    overlap_start = max(ver_min_age, perez_min_age)
    overlap_end = min(ver_max_age, perez_max_age)

    if overlap_end >= overlap_start:
        candidate_stints.append({
            'Stint': stint,
            'Compound': compound.iloc[0],
            'Laps': len(group),
            'OverlapStart': overlap_start,
            'OverlapEnd': overlap_end,
            'OverlapRange': overlap_end - overlap_start
        })

candidates = pd.DataFrame(candidate_stints)

if candidates.empty:
    raise ValueError(
        "No comparable Pérez stint found on the same compound."
    )

# Select the stint with the largest tyre-age overlap
best_candidate = candidates.sort_values(
    ['OverlapRange', 'Laps'],
    ascending=False
).iloc[0]

perez_stint_number = best_candidate['Stint']

perez_reference = perez_laps[
    perez_laps['Stint'] == perez_stint_number
].sort_values('TyreLife')

# Determine the tyre-age range available to both drivers
common_start = int(np.ceil(best_candidate['OverlapStart']))
common_end = int(np.floor(best_candidate['OverlapEnd']))

target_ages = [
    age for age in [3, 10, 18]
    if common_start <= age <= common_end
]

if len(target_ages) < 2:
    raise ValueError(
        "Not enough overlapping tyre ages for a useful comparison."
    )

# Select the closest available lap for each target tyre age
matched_laps = []

for age in target_ages:

    ver_lap = ver_reference.iloc[
        (ver_reference['TyreLife'] - age).abs().argmin()
    ]

    per_lap = perez_reference.iloc[
        (perez_reference['TyreLife'] - age).abs().argmin()
    ]

    matched_laps.append({
        'TargetAge': age,
        'VER': ver_lap,
        'PER': per_lap
    })

# Retrieve telemetry and calculate performance metrics
driver_telemetry = {}
performance_results = []

for match in matched_laps:

    for driver in ['VER', 'PER']:

        lap = match[driver]
        tyre_age = int(lap['TyreLife'])

        telemetry = lap.get_telemetry().copy()
        telemetry = telemetry.sort_values('Distance')

        driver_telemetry[(driver, match['TargetAge'])] = telemetry

        performance_results.append({
            'Driver': driver,
            'TargetTyreAge': match['TargetAge'],
            'ActualTyreAge': tyre_age,
            'LapNumber': int(lap['LapNumber']),
            'LapTimeSeconds': lap['LapTime'].total_seconds(),
            'AverageSpeed': telemetry['Speed'].mean(),
            'MaximumSpeed': telemetry['Speed'].max(),
            'MinimumSpeed': telemetry['Speed'].min(),
            'AverageThrottle': telemetry['Throttle'].mean(),
            'BrakingPercentage': (
                telemetry['Brake'].astype(bool).mean() * 100
            )
        })

performance_comparison = pd.DataFrame(performance_results)

print("COMPARISON DETAILS")
print("Compound:", reference_compound)
print("Verstappen stint: 1")
print("Pérez stint:", perez_stint_number)

print("\nMATCHED LAP PERFORMANCE")
display(performance_comparison.round(2))

# Plot speed traces for both drivers at matching tyre ages
fig, axes = plt.subplots(
    len(matched_laps), 1,
    figsize=(15, 4 * len(matched_laps)),
    sharex=True
)

if len(matched_laps) == 1:
    axes = [axes]

for ax, match in zip(axes, matched_laps):

    for driver in ['VER', 'PER']:

        telemetry = driver_telemetry[
            (driver, match['TargetAge'])
        ]

        ax.plot(
            telemetry['Distance'],
            telemetry['Speed'],
            label=driver
        )

    ax.set_title(
        f"Tyre Age {match['TargetAge']} | "
        f"{reference_compound}"
    )
    ax.set_ylabel("Speed (km/h)")
    ax.grid(True, alpha=0.3)
    ax.legend()

axes[-1].set_xlabel("Track Distance (m)")

plt.suptitle(
    "Verstappen vs Pérez | Abu Dhabi 2022",
    fontsize=15
)

plt.tight_layout()
plt.show()
```

```python

# ============================================================
# F1 TYRE DEGRADATION & PERFORMANCE ANALYSIS
# FINAL ANALYSIS, GRAPHS AND DATA EXPORT
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path

# ------------------------------------------------------------
# 1. OUTPUT FOLDERS
# ------------------------------------------------------------

output_dir = Path("outputs")
graph_dir = output_dir / "graphs"
table_dir = output_dir / "tables"

graph_dir.mkdir(parents=True, exist_ok=True)
table_dir.mkdir(parents=True, exist_ok=True)

print("Output folders ready.")


# ------------------------------------------------------------
# 2. PREPARE MULTI-RACE DATA
# ------------------------------------------------------------

final_data = multi_race_laps.copy()

# Remove old calculated columns to allow safe reruns
final_data = final_data.drop(
    columns=[
        "StintID",
        "BaselineLapTime",
        "LapTimeChange"
    ],
    errors="ignore"
)

final_data["StintID"] = (
    final_data["Event"].astype(str) + "_" +
    final_data["Driver"].astype(str) + "_" +
    final_data["Stint"].astype(str)
)

final_data = final_data.sort_values(
    ["Event", "Driver", "Stint", "TyreLife"]
).copy()

# Calculate baseline from first three recorded laps
baseline_data = (
    final_data.groupby("StintID")
    .head(3)
    .groupby("StintID")["LapTimeSeconds"]
    .mean()
    .rename("BaselineLapTime")
)

# Assign baseline to each lap
final_data["BaselineLapTime"] = (
    final_data["StintID"].map(baseline_data)
)

final_data["LapTimeChange"] = (
    final_data["LapTimeSeconds"] -
    final_data["BaselineLapTime"]
)

print("\nFinal dataset:", len(final_data), "laps")
print("Events:", final_data["Event"].nunique())
print("Stints:", final_data["StintID"].nunique())


# ------------------------------------------------------------
# 3. DATASET SUMMARY
# ------------------------------------------------------------

dataset_summary = pd.DataFrame({
    "Metric": [
        "Total Laps",
        "Total Stints",
        "Number of Events",
        "Number of Drivers",
        "Number of Compounds",
        "Average Lap Time"
    ],
    "Value": [
        len(final_data),
        final_data["StintID"].nunique(),
        final_data["Event"].nunique(),
        final_data["Driver"].nunique(),
        final_data["Compound"].nunique(),
        round(final_data["LapTimeSeconds"].mean(), 3)
    ]
})

print("\nDATASET SUMMARY")
display(dataset_summary)

dataset_summary.to_csv(
    table_dir / "dataset_summary.csv",
    index=False
)


# ------------------------------------------------------------
# 4. GRAPH: LAP TIME VS TYRE AGE
# VERSTAPPEN - ABU DHABI STINT 1
# ------------------------------------------------------------

ver_stint = final_data[
    (final_data["Event"] == "Abu Dhabi") &
    (final_data["Driver"] == "VER") &
    (final_data["Stint"] == 1)
].sort_values("TyreLife")

if not ver_stint.empty:

    plt.figure(figsize=(15, 5))

    plt.plot(
        ver_stint["TyreLife"],
        ver_stint["LapTimeSeconds"],
        marker="o",
        label="Verstappen"
    )

    plt.xlabel("Tyre Age (Laps)")
    plt.ylabel("Lap Time (Seconds)")
    plt.title("Lap Time vs Tyre Age - Verstappen, Abu Dhabi 2022")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()

    plt.savefig(
        graph_dir / "01_lap_time_vs_tyre_age.png",
        dpi=300
    )

    plt.show()


# ------------------------------------------------------------
# 5. COMPOUND PERFORMANCE SUMMARY
# ------------------------------------------------------------

compound_summary = (
    final_data.groupby(
        ["Event", "Compound", "TyreLife"]
    )
    .agg(
        MedianLapTimeChange=("LapTimeChange", "median"),
        LapSamples=("LapTimeChange", "count"),
        Stints=("StintID", "nunique"),
        Drivers=("Driver", "nunique")
    )
    .reset_index()
)

reliable_compound_data = compound_summary[
    (compound_summary["LapSamples"] >= 5) &
    (compound_summary["Stints"] >= 3)
].copy()

print("\nCOMPOUND PERFORMANCE SUMMARY")
display(reliable_compound_data)

compound_summary.to_csv(
    table_dir / "compound_performance_all.csv",
    index=False
)

reliable_compound_data.to_csv(
    table_dir / "compound_performance_reliable.csv",
    index=False
)


# ------------------------------------------------------------
# 6. GRAPH: TYRE AGE VS LAP TIME CHANGE
# SEPARATE GRAPHS FOR EACH EVENT
# ------------------------------------------------------------

for event in reliable_compound_data["Event"].unique():

    event_data = reliable_compound_data[
        reliable_compound_data["Event"] == event
    ]

    plt.figure(figsize=(15, 5))

    for compound in event_data["Compound"].unique():

        compound_data = event_data[
            event_data["Compound"] == compound
        ]

        plt.plot(
            compound_data["TyreLife"],
            compound_data["MedianLapTimeChange"],
            marker="o",
            label=compound
        )

    plt.xlabel("Tyre Age (Laps)")
    plt.ylabel("Median Lap Time Change (Seconds)")
    plt.title(f"Tyre Performance vs Tyre Age - {event} 2022")
    plt.grid(True)
    plt.legend(title="Compound")
    plt.tight_layout()

    filename = (
        event.lower().replace(" ", "_") +
        "_compound_performance.png"
    )

    plt.savefig(
        graph_dir / filename,
        dpi=300
    )

    plt.show()


# ------------------------------------------------------------
# 7. STINT DEGRADATION REGRESSION
# ------------------------------------------------------------

stint_results = []

for (event, driver, stint, compound), group in final_data.groupby(
    ["Event", "Driver", "Stint", "Compound"]
):

    group = group.sort_values("TyreLife")

    if (
        len(group) < 8 or
        group["TyreLife"].nunique() < 8
    ):
        continue

    x = group["TyreLife"].to_numpy()
    y = group["LapTimeSeconds"].to_numpy()

    slope, intercept = np.polyfit(x, y, 1)

    predicted = slope * x + intercept

    ss_res = np.sum((y - predicted) ** 2)
    ss_tot = np.sum((y - np.mean(y)) ** 2)

    r_squared = (
        1 - ss_res / ss_tot
        if ss_tot != 0 else np.nan
    )

    stint_results.append({
        "Event": event,
        "Driver": driver,
        "Stint": stint,
        "Compound": compound,
        "Laps": len(group),
        "DegradationRate": slope,
        "R2": r_squared
    })

degradation_rates = pd.DataFrame(stint_results)

degradation_summary = (
    degradation_rates.groupby(
        ["Event", "Compound"]
    )
    .agg(
        Stints=("DegradationRate", "count"),
        MedianRate=("DegradationRate", "median"),
        MeanRate=("DegradationRate", "mean"),
        MinimumRate=("DegradationRate", "min"),
        MaximumRate=("DegradationRate", "max"),
        MedianR2=("R2", "median")
    )
    .reset_index()
)

print("\nDEGRADATION RATE SUMMARY")
display(degradation_summary.round(4))

degradation_rates.to_csv(
    table_dir / "individual_stint_regressions.csv",
    index=False
)

degradation_summary.to_csv(
    table_dir / "degradation_rate_summary.csv",
    index=False
)


# ------------------------------------------------------------
# 8. GRAPH: DEGRADATION RATE BY COMPOUND AND EVENT
# ------------------------------------------------------------

for event in degradation_rates["Event"].unique():

    event_data = degradation_rates[
        degradation_rates["Event"] == event
    ]

    plt.figure(figsize=(15, 5))

    event_data.boxplot(
        column="DegradationRate",
        by="Compound",
        grid=True
    )

    plt.xlabel("Tyre Compound")
    plt.ylabel("Lap Time Trend (Seconds per Tyre-Age Lap)")
    plt.title(f"Stint Degradation Rate - {event} 2022")
    plt.suptitle("")
    plt.tight_layout()

    plt.savefig(
        graph_dir / (
            event.lower().replace(" ", "_") +
            "_degradation_rates.png"
        ),
        dpi=300
    )

    plt.show()


# ------------------------------------------------------------
# 9. WEATHER SUMMARY
# ------------------------------------------------------------

weather_columns = [
    "AirTemp",
    "TrackTemp",
    "Humidity",
    "Pressure",
    "WindSpeed"
]

weather_summary = (
    weather_data.groupby("Event")[weather_columns]
    .agg(["mean", "min", "max"])
    .round(2)
)

print("\nWEATHER SUMMARY")
display(weather_summary)

weather_summary.to_csv(
    table_dir / "weather_summary.csv"
)


# ------------------------------------------------------------
# 10. GRAPH: TRACK TEMPERATURE COMPARISON
# ------------------------------------------------------------

track_temperature = (
    weather_data.groupby("Event")["TrackTemp"]
    .mean()
)

plt.figure(figsize=(15, 5))

track_temperature.plot(
    kind="bar",
    rot=0
)

plt.xlabel("Grand Prix")
plt.ylabel("Average Track Temperature (°C)")
plt.title("Average Track Temperature - Abu Dhabi vs Bahrain 2022")
plt.grid(axis="y")
plt.tight_layout()

plt.savefig(
    graph_dir / "track_temperature_comparison.png",
    dpi=300
)

plt.show()


# ------------------------------------------------------------
# 11. VERSTAPPEN TELEMETRY GRAPHS
# ------------------------------------------------------------

if "telemetry_data" in globals() and telemetry_data:

    telemetry_summary = []

    for age, data in sorted(telemetry_data.items()):

        telemetry_summary.append({
            "TyreAge": age,
            "AverageSpeed": data["Speed"].mean(),
            "MaximumSpeed": data["Speed"].max(),
            "MinimumSpeed": data["Speed"].min(),
            "AverageThrottle": data["Throttle"].mean(),
            "BrakingPercentage": (
                (data["Brake"] > 0).mean() * 100
            ),
            "LapDistance": data["Distance"].max()
        })

    telemetry_summary = pd.DataFrame(telemetry_summary)

    print("\nVERSTAPPEN TELEMETRY SUMMARY")
    display(telemetry_summary.round(2))

    telemetry_summary.to_csv(
        table_dir / "verstappen_telemetry_summary.csv",
        index=False
    )

    for age, data in sorted(telemetry_data.items()):

        # SPEED
        plt.figure(figsize=(15, 5))

        plt.plot(
            data["Distance"],
            data["Speed"],
            label=f"Tyre Age {age}"
        )

        plt.xlabel("Distance (m)")
        plt.ylabel("Speed (km/h)")
        plt.title(f"Speed vs Distance - Verstappen, Tyre Age {age}")
        plt.grid(True)
        plt.legend()
        plt.tight_layout()

        plt.savefig(
            graph_dir / f"verstappen_speed_age_{age}.png",
            dpi=300
        )

        plt.show()

        # THROTTLE
        plt.figure(figsize=(15, 5))

        plt.plot(
            data["Distance"],
            data["Throttle"],
            label=f"Tyre Age {age}"
        )

        plt.xlabel("Distance (m)")
        plt.ylabel("Throttle (%)")
        plt.title(f"Throttle vs Distance - Verstappen, Tyre Age {age}")
        plt.grid(True)
        plt.legend()
        plt.tight_layout()

        plt.savefig(
            graph_dir / f"verstappen_throttle_age_{age}.png",
            dpi=300
        )

        plt.show()

        # BRAKE
        plt.figure(figsize=(15, 5))

        plt.plot(
            data["Distance"],
            data["Brake"],
            label=f"Tyre Age {age}"
        )

        plt.xlabel("Distance (m)")
        plt.ylabel("Brake")
        plt.title(f"Braking vs Distance - Verstappen, Tyre Age {age}")
        plt.grid(True)
        plt.legend()
        plt.tight_layout()

        plt.savefig(
            graph_dir / f"verstappen_braking_age_{age}.png",
            dpi=300
        )

        plt.show()


# ------------------------------------------------------------
# 12. EARLY VS LATE STINT SPEED COMPARISON
# ------------------------------------------------------------

if "telemetry_data" in globals() and len(telemetry_data) >= 2:

    early_age = min(telemetry_data.keys())
    late_age = max(telemetry_data.keys())

    early = telemetry_data[early_age]
    late = telemetry_data[late_age]

    common_distance = np.linspace(
        max(early["Distance"].min(), late["Distance"].min()),
        min(early["Distance"].max(), late["Distance"].max()),
        1000
    )

    early_speed = np.interp(
        common_distance,
        early["Distance"],
        early["Speed"]
    )

    late_speed = np.interp(
        common_distance,
        late["Distance"],
        late["Speed"]
    )

    speed_difference = late_speed - early_speed

    speed_comparison = pd.DataFrame({
        "Distance": common_distance,
        "EarlySpeed": early_speed,
        "LateSpeed": late_speed,
        "SpeedDifference": speed_difference
    })

    print("\nEARLY VS LATE STINT SPEED")
    print(
        "Average speed difference:",
        round(speed_difference.mean(), 2),
        "km/h"
    )

    print(
        "Largest speed loss:",
        round(speed_difference.min(), 2),
        "km/h"
    )

    print(
        "Largest speed gain:",
        round(speed_difference.max(), 2),
        "km/h"
    )

    speed_comparison.to_csv(
        table_dir / "early_vs_late_speed.csv",
        index=False
    )

    plt.figure(figsize=(15, 5))

    plt.plot(
        common_distance,
        speed_difference,
        label="Late tyre age - Early tyre age"
    )

    plt.axhline(0, linestyle="--")

    plt.xlabel("Distance (m)")
    plt.ylabel("Speed Difference (km/h)")
    plt.title("Speed Difference Through the Lap - Verstappen")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()

    plt.savefig(
        graph_dir / "early_vs_late_speed_difference.png",
        dpi=300
    )

    plt.show()


# ------------------------------------------------------------
# 13. VERSTAPPEN VS PÉREZ TELEMETRY COMPARISON
# ------------------------------------------------------------

if "driver_telemetry" in globals() and driver_telemetry:

    comparison_results = []

    available_ages = sorted(
        set(age for driver, age in driver_telemetry.keys())
    )

    for age in available_ages:

        if (
            ("VER", age) not in driver_telemetry or
            ("PER", age) not in driver_telemetry
        ):
            continue

        ver = driver_telemetry[("VER", age)]
        per = driver_telemetry[("PER", age)]

        common_distance = np.linspace(
            max(ver["Distance"].min(), per["Distance"].min()),
            min(ver["Distance"].max(), per["Distance"].max()),
            1000
        )

        ver_speed = np.interp(
            common_distance,
            ver["Distance"],
            ver["Speed"]
        )

        per_speed = np.interp(
            common_distance,
            per["Distance"],
            per["Speed"]
        )

        difference = ver_speed - per_speed

        comparison_results.append({
            "TargetTyreAge": age,
            "AverageSpeedDifference": difference.mean(),
            "MaximumVERAdvantage": difference.max(),
            "MaximumPERAdvantage": difference.min()
        })

        # SPEED TRACES
        plt.figure(figsize=(15, 5))

        plt.plot(
            common_distance,
            ver_speed,
            label="Verstappen"
        )

        plt.plot(
            common_distance,
            per_speed,
            label="Pérez"
        )

        plt.xlabel("Distance (m)")
        plt.ylabel("Speed (km/h)")
        plt.title(
            f"Speed Comparison - Verstappen vs Pérez, Tyre Age {age}"
        )

        plt.grid(True)
        plt.legend()
        plt.tight_layout()

        plt.savefig(
            graph_dir / f"driver_speed_comparison_age_{age}.png",
            dpi=300
        )

        plt.show()

        # SPEED DIFFERENCE
        plt.figure(figsize=(15, 5))

        plt.plot(
            common_distance,
            difference,
            label="Verstappen - Pérez"
        )

        plt.axhline(0, linestyle="--")

        plt.xlabel("Distance (m)")
        plt.ylabel("Speed Difference (km/h)")
        plt.title(
            f"Speed Difference - Verstappen vs Pérez, Tyre Age {age}"
        )

        plt.grid(True)
        plt.legend()
        plt.tight_layout()

        plt.savefig(
            graph_dir / f"driver_speed_difference_age_{age}.png",
            dpi=300
        )

        plt.show()

    if comparison_results:

        driver_speed_summary = pd.DataFrame(comparison_results)

        print("\nVERSTAPPEN VS PÉREZ SPEED SUMMARY")
        display(driver_speed_summary.round(2))

        driver_speed_summary.to_csv(
            table_dir / "verstappen_vs_perez_speed.csv",
            index=False
        )


# ------------------------------------------------------------
# 14. FINAL EXPORT SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 55)
print("FINAL ANALYSIS COMPLETE")
print("=" * 55)

print("\nGraphs saved in:", graph_dir.resolve())
print("Tables saved in:", table_dir.resolve())

print("\nGRAPH FILES:")
for file in sorted(graph_dir.glob("*.png")):
    print("-", file.name)

print("\nDATA FILES:")
for file in sorted(table_dir.glob("*.csv")):
    print("-", file.name)

print("\nAll available analysis has been exported.")
```
